```python
def conv2d_metrics(n_in, in_channels, n_filters, kernel_size, stride=1, padding=0):
    """
    Compute output size and number of parameters for a Conv2D layer.

    Parameters:
        n_in (int or tuple): input size (height, width)
        in_channels (int): number of input channels
        n_filters (int): number of output channels (filters)
        kernel_size (int or tuple): size of the kernel
        stride (int or tuple): stride
        padding (int or tuple): padding

    Returns:
        dict: output shape, total parameters, and formula breakdown
    """
    # Normalize inputs
    if isinstance(n_in, int):
        n_in = (n_in, n_in)
    if isinstance(kernel_size, int):
        kernel_size = (kernel_size, kernel_size)
    # if isinstance(stride, int):
    #     stride = (stride, stride)
    # if isinstance(padding, int):
    #     padding = (padding, padding)

    h_out = ((n_in[0] + 2 * padding - kernel_size[0]) // stride) + 1
    w_out = ((n_in[1] + 2 * padding - kernel_size[1]) // stride) + 1

    output_shape = (n_filters, h_out, w_out)
    num_params = n_filters *(in_channels * kernel_size[0] * kernel_size[1] + 1)  # +bias

    return {
        "Kernel Size": kernel_size,
        "Stride": stride,
        "Padding": padding,
        "Output Shape": output_shape,
        "Total Parameters": num_params
        # "Formula": f"(({n_in[0]} + 2×{padding[0]} - {kernel_size[0]}) / {stride[0]}) + 1 × (({n_in[1]} + 2×{padding[1]} - {kernel_size[1]}) / {stride[1]}) + 1",
        
        
    }
```

```python
def maxpool2d_metrics(input_size, kernel_size, stride=None, padding=0):
    """
    Compute output shape after MaxPool2D.

    Parameters:
        input_size (tuple): (H, W)
        kernel_size (int or tuple)
        stride (int or tuple): defaults to kernel_size
        padding (int or tuple)

    Returns:
        dict: output shape
    """
    if isinstance(kernel_size, int): kernel_size = (kernel_size, kernel_size)
    if stride is None: stride = kernel_size
    # if isinstance(stride, int): stride = (stride, stride)
    # if isinstance(padding, int): padding = (padding, padding)

    h_out = ((input_size[0] + 2 * padding - kernel_size[0]) // stride) + 1
    w_out = ((input_size[1] + 2 * padding - kernel_size[1]) // stride) + 1

    return {
        "Output Shape": (h_out, w_out),
        "Parameters": 0
    }
```

```python
def flatten_metrics(input_shape):
    """
    Compute flattened size from a 3D tensor.

    Parameters:
        input_shape (tuple): (C, H, W)

    Returns:
        dict: flattened size
    """
    c, h, w = input_shape
    return {
        "Flattened Size": c * h * w,
        "Output Shape": (c * h * w,)
    }
```
